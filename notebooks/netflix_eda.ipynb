```python
#13. How many unique genres are available in the dataset?
#As "Listed_in" attribute represents genres, each movie could have more than one genres
#Lets create an empty set to store unique genres after sperating from each record.
#i choose set as set can store only unique values
unique_genres = set()

# Loop through each row in the 'listed_in' column
for genres in df['listed_in']:
    unique_genres.update(genres.split(', '))  # Split and add genres to the set

# Print the unique genres and their count
print("Unique Genres:", unique_genres)
print("Total number of unique genres:", len(unique_genres))
```

```python
#14. How many movies and TV shows have been directed by each director?
director_counts = df.groupby('director').size()
print(director_counts)
```

```python
#15. How many shows/movies were added to Netflix each year?
df['year_added'] = df['date_added'].dt.year
year_counts = df.groupby('year_added').size()
print(year_counts)
```
